## Task 1: average

Write a function `average(numbers)` that returns the average of a list of numbers. For an empty list, return `0`.

```python
def average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)
```

## Task 2: top three

The list `scores` holds exam scores. Store the three highest scores in a new list `top_three`, highest first. Do not change `scores`.

```python
scores = [72, 95, 64, 88, 91, 57, 80]

top_three = sorted(scores, reverse=True)[:3]
```
